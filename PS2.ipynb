```python
from datasets import load_dataset
import pandas as pd
import matplotlib.pyplot as plt
```

#Load Dataset

```python
dataset = load_dataset("Cleanlab/student-grades")

df = pd.DataFrame(dataset["train"])
df.head()
```

#Check null values and shape of data

```python
print(df.shape)
df.isnull().sum()
```

#dropping unnecessary columns

```python
df.drop(['notes','noisy_letter_grade'],axis=1,inplace=True)
df.head()
```

#Calculating and creating average_score column

```python
df["average_score"] = (df["exam_1"] + df["exam_2"] + df["exam_3"]) / 3

print("Overall Average Score:",df["average_score"].mean())

df.head()
```

#Only first 10 students average score data is showing here

```python
plt.figure(figsize=(12, 5))

plt.bar(
    df["stud_ID"][:10],
    df["average_score"][:10]
)

plt.xlabel("Student ID")
plt.ylabel("Average Score")
plt.title("Average Score of Students")

plt.show()
```
